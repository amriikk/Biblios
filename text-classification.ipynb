```python
import pandas as pd
import numpy as np
from transformers import pipeline
from tqdm import tqdm

# Load the dataset updated from the previous step
books = pd.read_csv('books_cleaned.csv')

# Initialize the emotion classifier
emotion_classifier = pipeline("text-classification", model="j-hartmann/emotion-english-distilroberta-base", top_k=None, device="mps")

emotion_labels = ['anger', 'disgust', 'fear', 'joy', 'neutral', 'sadness', 'surprise']
isbn_list = []
emotion_scores = {label: [] for label in emotion_labels}

def calculate_max_emotion_scores(predictions):
    """Extracts the maximum probability for each emotion across all sentences in a description."""
    max_scores = {label: 0.0 for label in emotion_labels}
    for sentence_preds in predictions:
        for pred in sentence_preds:
            if pred['score'] > max_scores[pred['label']]:
                max_scores[pred['label']] = pred['score']
    return max_scores

# Loop through all books to extract emotional tone
for i in tqdm(range(len(books))):
    isbn_list.append(books['isbn13'][i])
    desc = str(books['description'][i])
    
    # Split description into individual sentences
    sentences = [s.strip() for s in desc.split('.') if s.strip()]
    
    if sentences:
        preds = emotion_classifier(sentences)
        max_scores = calculate_max_emotion_scores(preds)
    else:
        max_scores = {label: 0.0 for label in emotion_labels}
        
    for label in emotion_labels:
        emotion_scores[label].append(max_scores[label])

# Convert emotion scores to a dataframe and merge
emotions_df = pd.DataFrame(emotion_scores)
emotions_df['isbn13'] = isbn_list
books = pd.merge(books, emotions_df, on='isbn13')

# Save the final dataset before the Gradio Dashboard step
books.to_csv('books_final.csv', index=False)
```
